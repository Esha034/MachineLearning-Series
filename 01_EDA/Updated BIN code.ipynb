```python
import requests
import pandas as pd
import random
import time

url = "https://api.iinapi.com/iin"
api_key = "0fFcib8yt7vbLMzwXoh11X5Njy8jBknN"

# Realistic starting BIN prefixes (6 digits)
prefixes = ["424242", "400005", "555555", "510510", "601100", "353011", "356600", "378282", "371449", "620000", "220000"]

all_results = []

# Run for N random BINs
for _ in range(5000):  # Increase number for more data
    # Choose a random prefix (6 digits exactly)
    bin_number = random.choice(prefixes)

    params = {
        "key": api_key,
        "digits": bin_number   # Use exactly 6-digit prefix
    }

    response = requests.get(url, params=params)
    data = response.json()

    if isinstance(data, dict) and data.get("valid") and "result" in data:
        result = data["result"]
        all_results.append({
            "Bin": result.get("Bin"),
            "CardBrand": result.get("CardBrand"),
            "IssuingInstitution": result.get("IssuingInstitution"),
            "CardType": result.get("CardType"),
            "CardCategory": result.get("CardCategory"),
            "IssuingCountry": result.get("IssuingCountry"),
            "IssuingCountryCode": result.get("IssuingCountryCode")
        })
        print("Collected:", bin_number)
    else:
        print(f"Skipped invalid or unexpected response for BIN {bin_number}: {data}")

    time.sleep(1.5)  # API rate limit delay

# Convert to DataFrame and save CSV
df = pd.DataFrame(all_results)
df.to_csv("random_bin_dataset.csv", index=False)
print("Saved random BIN data to 'random_bin_dataset.csv'")
```
