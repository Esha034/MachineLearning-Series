# For Numerical-Numerical

## Scatterplot

```python
plt.figure(figsize=(6, 5))
sns.scatterplot(data=tips,
                x='total_bill',
                y='tip',
                hue='sex',
                style='smoker',
                size='size',
                # palette='cool',
                # sizes=(20, 200),
                alpha=1)

plt.title('Tip Amount vs Total Bill by Sex, Smoker & Size')
```

## B.Lineplot

```python
# we use lineplot when we have time based number in x axis
new =flight.groupby('year')[['passengers']].sum().reset_index()
sns.lineplot(data=new, x='year', y='passengers')
```

```python
#here clustermap is used to understand which month are behaving similar
sns.clustermap(flight.pivot_table(values='passengers',index='month',columns='year'))
```

# 2.Numerical -Categorical

## Bar plot

```python
titanic.head()
```

```python
# Seaborn’s barplot() by default shows the mean of the y variable, grouped by x,
# and can be split further using hue
plt.figure(figsize=(6, 5))
sns.barplot(data=titanic,
            x='Pclass',
            y='Age',
            hue='Sex',
            # palette='pastel',
            errorbar='sd')  # shows standard deviation

plt.title(" Average Age by Pclass and Sex on Titanic")
plt.ylabel("Average Age")
plt.xlabel("Passenger Class")
plt.show()
```

## Box plot

```python
sns.boxplot(data=tips,x='sex',y='tip',hue='smoker')
```

## Distplot

```python
# KDE work well with categorical comparisons
sns.kdeplot(data=titanic, x='Age', hue='Survived',fill=True)
```

# 3.Categorical-Categorical

## Heatmap

```python
pd.crosstab(titanic['Parch'],titanic['Survived'])
```

```python
ct = pd.crosstab(titanic['Pclass'], titanic['Survived'])
sns.heatmap(ct,annot=True, cmap='coolwarm')
```

## 6 Clustermap

```python
pd.crosstab(titanic['SibSp'], titanic['Survived'])
```

```python
sns.clustermap(pd.crosstab(titanic['Parch'], titanic['Survived']))
```
