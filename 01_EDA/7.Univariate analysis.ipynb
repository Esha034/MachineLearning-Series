# 1. EDA of Categorical Data

## A.Countplot

```python
df1['Pclass'].value_counts().plot(kind='bar')#displays the frequency of each category of the categorical column(here column =Pclass)
```

```python
df1['Survived'].value_counts().plot(kind='pie',autopct='%.2f')#displays the proportion of each category in terms of percentage
```
