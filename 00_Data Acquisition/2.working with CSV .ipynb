# 3. Opening a csv file from an URL

```python
import requests
from io import StringIO

# URL of the CSV file
url = "https://people.sc.fsu.edu/~jburkardt/data/csv/airtravel.csv"

# Send an HTTP GET request to the URL
response = requests.get(url)

# Check if the request was successful (status code 200)
if response.status_code == 200:
    # Use StringIO to read the content as a file-like object
    csv_data = StringIO(response.text)

    # Read the CSV data using pandas
    df = pd.read_csv(csv_data)
    # Display the first 5 rows
    print(df.head())
else:
    print("Failed to download CSV. Status code:", response.status_code) 
```

# 14.Coverters

```python
def rename(gender):
    if gender=="Male":
        return "M"
    else:
        return
```

```python
rename("Male")
```

```python
pd.read_csv('aug_train.csv.zip',converters={"gender":rename})
```

# 15 na_values parameter

## Why use na_values?
## Ensures consistency: All missing values are converted to NaN

## Makes it easier to filter, clean, or fill missing data

## Helpful when datasets use strings like "None", "NULL", "-", or "?" for missing entries

```python
pd.read_csv('aug_train.csv.zip',na_values=["None"])
#na_values parameter tells pandas to treat certain strings as missing values (NaN)
```

# 16. Loading a huge dataset in chunks

```python
pd.read_csv('BX-Books.csv.zip',sep=';',encoding='latin-1',on_bad_lines='skip')
```

```python
dfs=pd.read_csv('BX-Books.csv.zip',sep=';',encoding='latin-1',on_bad_lines='skip',chunksize=90000)
```
