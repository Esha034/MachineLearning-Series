```python
import pandas as pd 
import requests
from bs4 import BeautifulSoup
```

```python
#we have to send the requests as we are sending it from browser(this request is coming from browser)

# Set the headers to simulate a browser visit
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'
}

# Correct URL
url = 'https://www.ambitionbox.com/list-of-companies?page=1'

# Send request
webpage = requests.get(url, headers=headers).text
```

Now we will pass this webpage to beautiful soup to apply methods of beautiful soup.

```python
soup=BeautifulSoup(webpage,'lxml')
```

```python
print(soup.prettify())
```
