```python
import pandas as pd
import numpy as np
```

```python
date=pd.read_csv('orders.csv')
time=pd.read_csv('messages.csv')
```

```python
date.drop(columns=['product_id','city_id','orders'],inplace=True)
date.head()
```

## Working with date dataset

```python
date.info()
```

```python
date['date']=pd.to_datetime(date['date'])
date.info()
```

```python
# to extract year
date['year'] = date['date'].dt.year
# to extract month
date['month'] = date['date'].dt.month
# to extract month name
date['date_month_name'] = date['date'].dt.month_name()
 # to extract date
date['day'] = date['date'].dt.day
#to extract date name
date['weekday'] = date['date'].dt.day_name()
# to extract dayofweek
date['date_dow'] = date['date'].dt.dayofweek
#to extract weekno. of the year
date['date_week'] = date['date'].dt.isocalendar().week
date.sample(5)
```

```python
#is weekend ?
date['date_is_weekend']=np.where(date['weekday'].isin(['Sunday','Saturday']),1,0)
date.head()
```

```python
date.info()
```

```python
date['quarter'] = date['date'].dt.quarter
date.sample()
```

```python
date['semester']=np.where(date['quarter'].isin([1,2]),1,2)
date.sample()
```

```python
#extract time elapse between dates
from datetime import datetime
today=datetime.today()
print(today)
```

```python
today.date()
```

```python
today.time()
```

```python
today-date['date']
```

## Months passed from today(Important)

```python
#Exact res
np.round(today.year - date['date'].dt.year) * 12 + (today.month - date['date'].dt.month)
```

```python
#approx result
np.round(((today - date['date']) / np.timedelta64(1, 'D')) / 30,0).astype(int)
# np.round(months_passed, 0)
```
